# Quantum Variational Classifier

Minimal package-client example for the PennyLane-based VQC workflow in `qml.classifiers`.

## Tutorial Goals

This tutorial walks through the minimal variational quantum classifier workflow. You will:

- call the high-level `run_vqc` API
- train a small angle-embedding classifier
- inspect accuracy, loss history, predictions, and returned keys
- see the expected shape of a package-client notebook

The notebook focuses on API usage rather than proving model advantage.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().resolve()
if not (repo_root / "src" / "qml").exists():
    repo_root = next(
        (parent for parent in repo_root.parents if (parent / "src" / "qml").exists()),
        repo_root,
    )
package_root = repo_root / "src"
if str(package_root) not in sys.path:
    sys.path.insert(0, str(package_root))
```

```python
from qml.classifiers import run_vqc
```

## Run a minimal VQC experiment

This trains a small variational quantum classifier on the two-moons dataset.

```python
result = run_vqc(
    n_samples=200,
    noise=0.1,
    test_size=0.25,
    seed=0,
    n_layers=2,
    steps=50,
    step_size=0.1,
    plot=True,
    save=True,
)
```

## Summary metrics

```python
print("Train accuracy:", result["train_accuracy"])
print("Test accuracy:", result["test_accuracy"])
print("Final loss:", result["final_loss"])
```

## First few loss values

```python
result["loss_history"][:10]
```

```python
sorted(result.keys())
```

## Notes

The notebook intentionally acts as a pure package client:

- dataset generation lives in `qml.data`
- embedding logic lives in `qml.embeddings`
- ansatz logic lives in `qml.ansatz`
- training/evaluation orchestration lives in `qml.classifiers`

This keeps the notebook focused on usage rather than implementation.
