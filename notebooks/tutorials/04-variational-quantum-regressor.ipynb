# Variational Quantum Regressor

Minimal package-client example for the PennyLane-based regression workflow in `qml.regression`.

## Tutorial Goals

This tutorial walks through the minimal variational quantum regressor workflow. You will:

- call the high-level `run_vqr` API
- fit a compact expectation-value regressor
- inspect regression metrics, predictions, and result keys
- use the output structure expected by downstream docs and examples

The notebook is intentionally small so it can be re-run while developing.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().resolve()
if not (repo_root / "src" / "qml").exists():
    repo_root = next(
        (parent for parent in repo_root.parents if (parent / "src" / "qml").exists()),
        repo_root,
    )
package_root = repo_root / "src"
if str(package_root) not in sys.path:
    sys.path.insert(0, str(package_root))
```

```python
from qml.regression import run_vqr
```

## Run a minimal variational regression experiment

```python
result = run_vqr(
    n_samples=100,
    noise=0.1,
    test_size=0.25,
    seed=0,
    n_layers=2,
    steps=50,
    step_size=0.1,
    plot=True,
    save=True,
)
```

## Summary metrics

```python
print("Train MSE:", result["train_mse"])
print("Test MSE:", result["test_mse"])
print("Train MAE:", result["train_mae"])
print("Test MAE:", result["test_mae"])
print("Final loss:", result["final_loss"])
```
