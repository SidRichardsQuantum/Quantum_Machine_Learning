# Quantum Kernel Classifier

Minimal package-client example for the PennyLane-based quantum kernel workflow in `qml.kernel_methods`.

## Tutorial Goals

This tutorial introduces the package-level quantum kernel classifier. You will:

- call `run_quantum_kernel_classifier`
- fit an SVM from a precomputed quantum fidelity kernel
- inspect the kernel matrix and held-out predictions
- understand which outputs are useful for diagnostics

Use this before the estimator tutorial if you want the packaged end-to-end workflow first.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().resolve()
if not (repo_root / "src" / "qml").exists():
    repo_root = next(
        (parent for parent in repo_root.parents if (parent / "src" / "qml").exists()),
        repo_root,
    )
package_root = repo_root / "src"
if str(package_root) not in sys.path:
    sys.path.insert(0, str(package_root))
```

```python
from qml.kernel_methods import run_quantum_kernel_classifier
```

## Run a minimal quantum kernel experiment

This fits a quantum kernel classifier on the two-moons dataset.

```python
result = run_quantum_kernel_classifier(
    n_samples=100,
    noise=0.1,
    test_size=0.25,
    seed=0,
    plot=True,
    save=True,
)
```

## Summary metrics

```python
print("Train accuracy:", result["train_accuracy"])
print("Test accuracy:", result["test_accuracy"])
```
